## E1.3 Pump curve

The pump head depends on the flowrate through a polynomial formula. The formula
is written for flowrate in liters per minute, so the head comes out in meters.

Pseudocode:
1. Start.
2. Ask for the flowrate Q in liters per minute.
3.    3. Calculate the head: `h = 31.2 + 0.003*Q - 4.29e-6*Q^2 - 8.6e-10*Q^3`
4. Show the head h in meters.
5. Stop.


```python
def pump_head(Q):
    # Q is the flowrate in L/min, the result is the head in meters
    return 31.2 + 0.003*Q - 4.29e-6*Q**2 - 8.6e-10*Q**3

Q = 500   # example flowrate in L/min
h = pump_head(Q)

print(f"Flowrate Q = {Q} L/min")
print(f"Head     h = {h:.4f} m")
```

## E1.34 Error of the surface-area estimate

The true area is 1.25 m^2 and the computer model estimated 1.252 m^2.
To see how far off the estimate is, we use three measures:

- Absolute error = |true value - estimated value|
- Relative error = absolute error / |true value|
- Percent relative error = relative error x 100


```python
A_true = 1.25    # true area in m^2
A_est  = 1.252   # estimated area in m^2

abs_err     = abs(A_true - A_est)     # how far off the estimate is
rel_err     = abs_err / abs(A_true)   # the error compared to the true value
pct_rel_err = rel_err * 100           # the same thing as a percent

print(f"Absolute error         = {abs_err:.4f} m^2")
print(f"Relative error         = {rel_err:.4f}")
print(f"Percent relative error = {pct_rel_err:.2f} %")
```

## E1.11 Four means of a sequence (without an array)

The numbers are read from a file one at a time, so we cannot store them in an
array. Instead we keep running totals while reading.

Pseudocode:
1. Start with n = 0 and all totals = 0
   (sum of x, sum of 1/x, sum of ln(x), sum of x^2).
2. Open the file.
3. While there is another number x in the file:
   - add 1 to n
   - add x to the sum
   - add 1/x to the sum of reciprocals
   - add ln(x) to the sum of logs
   - add x^2 to the sum of squares
4. Close the file.
5. Calculate the four means:
   - arithmetic mean = sum / n
   - harmonic mean = n / (sum of reciprocals)
   - geometric mean = exp(sum of logs / n)
   - quadratic mean = sqrt(sum of squares / n)
6. Show the four means and stop.


```python
# make a small test file with one number on each line
with open("data.txt", "w") as f:
    f.write("2\n4\n8\n16\n")

print("data.txt created")
```

```python
import math

n = 0
total = total_inv = total_log = total_sq = 0.0

# read the file one line at a time, so no array is needed
with open("data.txt") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue                      # skip empty lines
        x = float(line)
        n         += 1
        total     += x
        total_inv += 1/x
        total_log += math.log(x)
        total_sq  += x**2

arithmetic = total / n
harmonic   = n / total_inv
geometric  = math.exp(total_log / n)      # using logs avoids a huge product
quadratic  = math.sqrt(total_sq / n)

print(f"n = {n}")
print(f"Arithmetic mean = {arithmetic:.4f}")
print(f"Harmonic mean   = {harmonic:.4f}")
print(f"Geometric mean  = {geometric:.4f}")
print(f"Quadratic mean  = {quadratic:.4f}")
```
